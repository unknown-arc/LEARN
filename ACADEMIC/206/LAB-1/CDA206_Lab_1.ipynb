```python
import numpy as np

# Create an array that start with 1, end with 20, incremented by 3.
c=np.arange(1,20,2)
print(c)

lin_space=np.linspace(0,10,7)
print(lin_space)

lin_space2=np.linspace(0,10,7,dtype=int)
print(lin_space2)

vec_ones=np.ones(4)
print(vec_ones)

vec_zeros=np.zeros(4)
print(vec_zeros)

one_dim_vec=np.array([1,2,3,4,5,6])
print(one_dim_vec)

multi_dim=np.reshape(one_dim_vec,(3,2))
print(multi_dim)

# It gives the shape of an array
print(multi_dim.shape)

# Size=total number of elements in array
print(multi_dim.size)

# Define a matrix
A=np.array([[1,2],[3,4],[5,6]])
print(A)
```

```python
v1=np.array([1,2,3])
v2=np.array([6,2,4])

print(f"The addition of vectors {v1} and {v2} is {v1+v2}")
print(f"The substraction of vectors {v1} and {v2} is {v1-v2}")
print(f"The dot product of vectors {v1} and {v2} is {v1@v2}")
print(f"The cross  product of vectors {v1} and {v2} is {np.cross(v1,v2)}")
print(f"The magnitude of vector {v1} is {np.linalg.norm(v1)}")

print(f"The p-norm of vector {v1} is {np.linalg.norm(v1,ord=2)}")
```

```python
# V=[v1,v2,...,vn]
# p_norm=(|v1|**p+|v2|**p+...+|vn|**p)**(1/p)

def p_norm(v,p):
    return (np.sum(np.abs(v)**p))**(1/p)

v1=np.array([1,2,3])
p=int(input(f"Enter the choice of p"))
print(p_norm(v1,p))
```

```python
# The angle between two vectors
v1=np.array([1,1])
v2=np.array([2,2])

cos_theta=np.dot(v1,v2)/( p_norm(v1,2)* p_norm(v2,2))
theta=np.arccos(cos_theta) # It is in radian
print(np.degrees(theta))
```

```python
# Normization
v=np.array([-100,200,300])

normalization_v=v/p_norm(v,3)
print(normalization_v)
print(p_norm(normalization_v,3))
```

```python
A=np.array([[1,2],[3,4]])
B=np.array([[4,3],[2,1]])
print(f"The sum of two matrices is {A+B}")
print(f"Elementwise product of two matrices is {A*B}")
print(f"The product of two matrices is {A@B}")
```

```python
import sympy as sy
a,b,c,d,e,f,g,h,i,j,k,l=sy.symbols("a,b,c,d,e,f,g,h,i,j,k,l",real=True)

A=sy.Matrix([[a,b,c],[d,e,f]])
B=sy.Matrix([[g,h,i],[j,k,l]])
print(f"Symbolic addition {A+B}")
print(f"Symbolic multiplication {A@B.transpose()}")
```
