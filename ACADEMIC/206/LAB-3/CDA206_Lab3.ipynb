```python
# Determinant of matrix
import numpy as np

def determinant(matrix):

    rows, cols = matrix.shape

    if rows != cols:
        return "Determinant can't be evaluated"

    if rows == 1:
        return matrix[0, 0]

    if rows == 2:
        return matrix[0, 0] * matrix[1, 1] - matrix[0, 1] * matrix[1, 0]

    answer = 0

    for col in range(cols):

        minor = np.delete(matrix, 0, axis=0)
        minor = np.delete(minor, col, axis=1)
        answer = answer + ((-1)**col) * matrix[0, col] * determinant(minor)

    return answer
```

```python
matrix = np.array([[6, 1, 4], [9,1,4], [2, 8, 7]])
print(f"The determinant of matrix {determinant(matrix)}")
```

```python
# Determinant to find Eigen value
import numpy as np

def determinant_sp(matrix):

    n=len(matrix)

    # if rows != cols:
    #     return "Determinant can't be evaluated"

    if n == 1:
        return matrix[0][0]

    if n == 2:
        return matrix[0][0] * matrix[1][1] - matrix[0][1] * matrix[1][0]

    answer = 0

    for col in range(n):

        minor=[list(row[:col]) + list(row[col + 1:]) for row in matrix[1:]]
        answer = answer + ((-1)**col) * matrix[0][col] * determinant_sp(minor)

    return answer
```

```python
import sympy as sp
lamda=sp.symbols('lamda')

def charaterstic_poly(mat):
    n=len(mat)
    # mat-lam*I
    char_matrix=[[mat[i][j]-(lamda if i==j else 0) for j in range(n)] for i in range(n)]
    char_poly=determinant_sp(char_matrix)
    return sp.expand(char_poly)
```

```python
matrix1 = np.array([[6, 0, 0], [0,3,0], [0, 0, 7]])
print(charaterstic_poly(matrix1))
```

```python
sp.solvers.solve(charaterstic_poly(matrix1))
```
